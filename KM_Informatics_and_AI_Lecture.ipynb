# 1. Model Equation

Our neural network has only one dense neuron:

$$
\hat{y} = wx + b
$$

where:

- $x$ = input value  
- $w$ = weight  
- $b$ = bias  
- $\hat{y}$ = predicted value  

At **Epoch 1**, we start with:

$$
w = 0
$$

$$
b = 0
$$

So the prediction becomes:

$$
\hat{y} = 0 \times x + 0
$$

$$
\hat{y} = 0
$$

Therefore, for every input, the model predicts:

$$
\hat{y} = 0
$$

---

# 2. Loss Equation for One Sample

For one sample, the squared error loss is:

$$
L = (\hat{y} - y)^2
$$

Let:

$$
e = \hat{y} - y
$$

So:

$$
L = e^2
$$

where:

- $e$ = error  
- $\hat{y}$ = predicted value  
- $y$ = true value  

---

# 3. Derivative of Loss with Respect to Error

We have:

$$
L = e^2
$$

Differentiate with respect to $e$:

$$
\frac{\partial L}{\partial e} = 2e
$$

Since:

$$
e = \hat{y} - y
$$

we can write:

$$
\frac{\partial L}{\partial e} = 2(\hat{y} - y)
$$

---

# 4. Derivative of Error with Respect to Prediction

The error is:

$$
e = \hat{y} - y
$$

Differentiate with respect to $\hat{y}$:

$$
\frac{\partial e}{\partial \hat{y}} = 1
$$

Because:

$$
\frac{\partial}{\partial \hat{y}}(\hat{y}) = 1
$$

and:

$$
\frac{\partial}{\partial \hat{y}}(y) = 0
$$

So:

$$
\frac{\partial e}{\partial \hat{y}} = 1
$$

---

# 5. Derivative of Prediction with Respect to Bias

The prediction equation is:

$$
\hat{y} = wx + b
$$

Differentiate with respect to $b$:

$$
\frac{\partial \hat{y}}{\partial b}
=
\frac{\partial}{\partial b}(wx + b)
$$

Now split the derivative:

$$
\frac{\partial}{\partial b}(wx + b)
=
\frac{\partial}{\partial b}(wx)
+
\frac{\partial}{\partial b}(b)
$$

Since $wx$ does not contain $b$:

$$
\frac{\partial}{\partial b}(wx) = 0
$$

And:

$$
\frac{\partial}{\partial b}(b) = 1
$$

Therefore:

$$
\frac{\partial \hat{y}}{\partial b} = 0 + 1
$$

$$
\frac{\partial \hat{y}}{\partial b} = 1
$$

---

# 6. Apply the Chain Rule

We want:

$$
\frac{\partial L}{\partial b}
$$

The dependency path is:

$$
b \rightarrow \hat{y} \rightarrow e \rightarrow L
$$

So by the chain rule:

$$
\frac{\partial L}{\partial b}
=
\frac{\partial L}{\partial e}
\times
\frac{\partial e}{\partial \hat{y}}
\times
\frac{\partial \hat{y}}{\partial b}
$$

Now substitute each part:

$$
\frac{\partial L}{\partial e} = 2e
$$

$$
\frac{\partial e}{\partial \hat{y}} = 1
$$

$$
\frac{\partial \hat{y}}{\partial b} = 1
$$

So:

$$
\frac{\partial L}{\partial b}
=
2e \times 1 \times 1
$$

$$
\frac{\partial L}{\partial b}
=
2e
$$

Since:

$$
e = \hat{y} - y
$$

we get:

$$
\frac{\partial L}{\partial b}
=
2(\hat{y} - y)
$$

This is the bias gradient for **one sample**.

---

# 7. Bias Gradient for MSE with Many Samples

In our code, we use **Mean Squared Error**, not single-sample squared error.

The MSE loss is:

$$
L =
\frac{1}{N}
\sum_{i=1}^{N}
(\hat{y}_i - y_i)^2
$$

Now differentiate with respect to $b$:

$$
\frac{\partial L}{\partial b}
=
\frac{\partial}{\partial b}
\left[
\frac{1}{N}
\sum_{i=1}^{N}
(\hat{y}_i - y_i)^2
\right]
$$

The constant $\frac{1}{N}$ stays outside:

$$
\frac{\partial L}{\partial b}
=
\frac{1}{N}
\sum_{i=1}^{N}
\frac{\partial}{\partial b}
(\hat{y}_i - y_i)^2
$$

For each sample:

$$
\frac{\partial}{\partial b}
(\hat{y}_i - y_i)^2
=
2(\hat{y}_i - y_i)
$$

So:

$$
\frac{\partial L}{\partial b}
=
\frac{1}{N}
\sum_{i=1}^{N}
2(\hat{y}_i - y_i)
$$

Move $2$ outside:

$$
\frac{\partial L}{\partial b}
=
\frac{2}{N}
\sum_{i=1}^{N}
(\hat{y}_i - y_i)
$$

Since:

$$
error_i = \hat{y}_i - y_i
$$

the final formula becomes:

$$
\boxed{
\frac{\partial L}{\partial b}
=
\frac{2}{N}
\sum_{i=1}^{N}
error_i
}
$$

---

# 8. Apply Epoch 1 Values

In our data, we have 7 samples:

$$
N = 7
$$

At Epoch 1:

$$
w = 0
$$

$$
b = 0
$$

Therefore:

$$
\hat{y} = 0
$$

So the errors are:

$$
error_1 = 0 - (-40) = 40
$$

$$
error_2 = 0 - 14 = -14
$$

$$
error_3 = 0 - 32 = -32
$$

$$
error_4 = 0 - 46.4 = -46.4
$$

$$
error_5 = 0 - 59 = -59
$$

$$
error_6 = 0 - 71.6 = -71.6
$$

$$
error_7 = 0 - 100.4 = -100.4
$$

---

# 9. Sum the Errors

Now sum all errors:

$$
\sum error
=
40 + (-14) + (-32) + (-46.4) + (-59) + (-71.6) + (-100.4)
$$

Step-by-step:

$$
40 - 14 = 26
$$

$$
26 - 32 = -6
$$

$$
-6 - 46.4 = -52.4
$$

$$
-52.4 - 59 = -111.4
$$

$$
-111.4 - 71.6 = -183.0
$$

$$
-183.0 - 100.4 = -283.4
$$

So:

$$
\sum error = -283.4
$$

In TensorFlow output, this appears as:

$$
\sum error = -283.399994
$$

The tiny difference is due to floating-point precision.

---

# 10. Calculate the Bias Gradient

The formula is:

$$
\frac{\partial L}{\partial b}
=
\frac{2}{N}
\sum error
$$

Substitute the values:

$$
\frac{\partial L}{\partial b}
=
\frac{2}{7}
\times
(-283.399994)
$$

Multiply first:

$$
2 \times (-283.399994)
=
-566.799988
$$

Then divide by 7:

$$
\frac{-566.799988}{7}
=
-80.971426
$$

Therefore:

$$
\boxed{
\frac{\partial L}{\partial b}
=
-80.971428
}
$$

---

# 11. Update the Bias

Gradient descent update rule:

$$
b_{new}
=
b_{old}
-
\alpha
\frac{\partial L}{\partial b}
$$

where:

- $\alpha$ = learning rate  
- $b_{old}$ = previous bias  
- $b_{new}$ = updated bias  

In our code:

$$
\alpha = 0.0001
$$

$$
b_{old} = 0
$$

$$
\frac{\partial L}{\partial b} = -80.971428
$$

Substitute:

$$
b_{new}
=
0
-
0.0001(-80.971428)
$$

Negative times negative becomes positive:

$$
b_{new}
=
0
+
0.0080971428
$$

So:

$$
\boxed{
b_{new} = 0.008097
}
$$

---

# 12. Final Summary

The model is:

$$
\hat{y} = wx + b
$$

The loss is:

$$
L =
\frac{1}{N}
\sum
(\hat{y} - y)^2
$$

The error is:

$$
error = \hat{y} - y
$$

The bias gradient is:

$$
\frac{\partial L}{\partial b}
=
\frac{2}{N}
\sum error
$$

For Epoch 1:

$$
\sum error = -283.399994
$$

So:

$$
\frac{\partial L}{\partial b}
=
\frac{2}{7}
(-283.399994)
$$

$$
\frac{\partial L}{\partial b}
=
-80.971428
$$

Then the bias update is:

$$
b_{new}
=
b_{old}
-
lr
\times
\frac{\partial L}{\partial b}
$$

$$
b_{new}
=
0
-
0.0001(-80.971428)
$$

$$
b_{new}
=
0.008097
$$

---

# Simple Intuition

At Epoch 1, the model predicts:

$$
\hat{y}=0
$$

But most true Fahrenheit values are much larger than 0.

So the prediction is mostly too low.

That makes the total error negative:

$$
\sum error < 0
$$

Therefore:

$$
\frac{\partial L}{\partial b} < 0
$$

Gradient descent subtracts the gradient:

$$
b_{new} = b_{old} - lr \times gradient
$$

Since the gradient is negative, subtracting it increases the bias.

So the bias moves upward:

$$
0 \rightarrow 0.008097
$$

This is correct because the model needs to increase its predictions.

```python
input = [-40, -10,  0,  8, 15, 22,  38]
output = [-40.0, 14.0, 32.0, 46.4, 59.0, 71.6, 100.4]
```

```python
import tensorflow as tf
import numpy as np



l0 = tf.keras.layers.Dense(units=1, input_shape=[1])



model = tf.keras.Sequential([l0])



model.compile(loss='mean_squared_error',
              optimizer = tf.keras.optimizers.SGD(learning_rate=0.0001))


history = model.fit(np.array(input), np.array(output), epochs=1000, verbose=False)
print("These are the layer variables, weight:",l0.get_weights()[0],", bias:",l0.get_weights()[1])
```

```python
import tensorflow as tf
import numpy as np
import pandas as pd

# -----------------------------
# Data
# -----------------------------
x_np = np.array([-40, -10, 0, 8, 15, 22, 38], dtype=np.float32).reshape(-1, 1)
y_np = np.array([-40.0, 14.0, 32.0, 46.4, 59.0, 71.6, 100.4], dtype=np.float32).reshape(-1, 1)



x = tf.constant(x_np)
y = tf.constant(y_np)

N = len(x_np)


w = tf.Variable([[0]], dtype=tf.float32)
b = tf.Variable([0], dtype=tf.float32)


learning_rate = 0.0001
epochs = 10


all_epoch_rows = []

# -----------------------------
# Training loop
# -----------------------------
for epoch in range(1, epochs + 1):

    with tf.GradientTape() as tape:
        y_pred = tf.matmul(x, w) + b
        error = y_pred - y
        loss = tf.reduce_mean(tf.square(error))

    dw, db = tape.gradient(loss, [w, b])

    # Save old values before update
    old_w = w.numpy()[0, 0]
    old_b = b.numpy()[0]
    loss_value = loss.numpy()
    dw_value = dw.numpy()[0, 0]
    db_value = db.numpy()[0]

    # Manual update
    new_w = old_w - learning_rate * dw_value
    new_b = old_b - learning_rate * db_value

    # Apply update
    w.assign([[new_w]])
    b.assign([new_b])

    # Save summary row
    all_epoch_rows.append({
        "epoch": epoch,
        "w_before": old_w,
        "b_before": old_b,
        "loss": loss_value,
        "dL_dw": dw_value,
        "dL_db": db_value,
        "w_after": new_w,
        "b_after": new_b
    })

    # -----------------------------
    # First 3 epochs: full manual explanation
    # -----------------------------
    if epoch <= 3:
        print("\n" + "=" * 90)
        print(f"EPOCH {epoch} FULL BACKPROPAGATION CALCULATION")
        print("=" * 90)

        print("\nModel equation:")
        print("y_pred = w*x + b")

        print("\nLoss equation:")
        print("MSE = (1/N) * Σ(y_pred - y_true)^2")

        print("\nGradient equations:")
        print("dL/dw = (2/N) * Σ((y_pred - y_true) * x)")
        print("dL/db = (2/N) * Σ(y_pred - y_true)")

        print("\nCurrent parameters before update:")
        print(f"w = {old_w:.6f}")
        print(f"b = {old_b:.6f}")

        # Build detailed per-sample table
        detailed_rows = []

        sum_error_sq = 0.0
        sum_error_x = 0.0
        sum_error = 0.0

        for i in range(N):
            xi = float(x_np[i, 0])
            yi = float(y_np[i, 0])

            pred_i = old_w * xi + old_b
            err_i = pred_i - yi
            err_sq_i = err_i ** 2
            err_x_i = err_i * xi

            sum_error_sq += err_sq_i
            sum_error_x += err_x_i
            sum_error += err_i

            detailed_rows.append({
                "i": i + 1,
                "x": xi,
                "y_true": yi,
                "y_pred = w*x+b": pred_i,
                "error = y_pred-y": err_i,
                "error^2": err_sq_i,
                "error*x": err_x_i
            })

        df_detail = pd.DataFrame(detailed_rows)

        print("\nPer-sample forward calculation:")
        print(df_detail.to_string(index=False))

        manual_loss = sum_error_sq / N
        manual_dw = (2 / N) * sum_error_x
        manual_db = (2 / N) * sum_error

        print("\nManual loss calculation:")
        print(f"Σ(error^2) = {sum_error_sq:.6f}")
        print(f"MSE = Σ(error^2) / N = {sum_error_sq:.6f} / {N}")
        print(f"MSE = {manual_loss:.6f}")

        print("\nManual gradient calculation for weight:")
        print(f"Σ(error*x) = {sum_error_x:.6f}")
        print(f"dL/dw = (2/N) * Σ(error*x)")
        print(f"dL/dw = (2/{N}) * {sum_error_x:.6f}")
        print(f"dL/dw = {manual_dw:.6f}")

        print("\nManual gradient calculation for bias:")
        print(f"Σ(error) = {sum_error:.6f}")
        print(f"dL/db = (2/N) * Σ(error)")
        print(f"dL/db = (2/{N}) * {sum_error:.6f}")
        print(f"dL/db = {manual_db:.6f}")

        print("\nParameter update equations:")
        print("w_new = w_old - learning_rate * dL/dw")
        print("b_new = b_old - learning_rate * dL/db")

        print("\nManual parameter update:")
        print(f"w_new = {old_w:.6f} - {learning_rate} * ({manual_dw:.6f})")
        print(f"w_new = {new_w:.6f}")

        print(f"b_new = {old_b:.6f} - {learning_rate} * ({manual_db:.6f})")
        print(f"b_new = {new_b:.6f}")

# -----------------------------
# Final table for all epochs
# -----------------------------
df_epochs = pd.DataFrame(all_epoch_rows)

print("\n" + "=" * 90)
print("SUMMARY TABLE FOR ALL EPOCHS")
print("=" * 90)

print(df_epochs.to_string(index=False))

print("\nFinal learned parameters:")
print(f"w = {w.numpy()[0, 0]:.6f}")
print(f"b = {b.numpy()[0]:.6f}")
```
