```python
import tensorflow as tf
import numpy as np

x = np.array([-40, -10, 0, 8, 15, 22, 38], dtype=np.float32)
y = np.array([-40.0, 14.0, 32.0, 46.4, 59.0, 71.6, 100.4], dtype=np.float32)

x = x.reshape(-1, 1)
y = y.reshape(-1, 1)

l0 = tf.keras.layers.Dense(units=1, input_shape=[1])
model = tf.keras.Sequential([l0])

optimizer = tf.keras.optimizers.SGD(learning_rate=0.0001)

for epoch in range(10):
    with tf.GradientTape() as tape:
        y_pred = model(x, training=True)
        loss = tf.reduce_mean(tf.square(y_pred - y))

    gradients = tape.gradient(loss, model.trainable_variables)

    w = model.trainable_variables[0]
    b = model.trainable_variables[1]

    grad_w = gradients[0]
    grad_b = gradients[1]

    print(f"\nEpoch {epoch+1}")
    print("Current weight:", w.numpy())
    print("Current bias:", b.numpy())

    print("Predictions:")
    print(y_pred.numpy().reshape(-1))

    print("Errors:")
    print((y_pred - y).numpy().reshape(-1))

    print("Loss:", loss.numpy())

    print("Gradient dL/dw:", grad_w.numpy())
    print("Gradient dL/db:", grad_b.numpy())

    optimizer.apply_gradients(zip(gradients, model.trainable_variables))

    print("Updated weight:", w.numpy())
    print("Updated bias:", b.numpy())
```
