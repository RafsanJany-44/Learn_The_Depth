# Manual Backpropagation Calculation for Weight and Bias for <font color = 'blue'>EPOCH 1</font>
We want to understand how the model calculates both gradients:

$$
\frac{\partial L}{\partial w}
$$

and

$$
\frac{\partial L}{\partial b}
$$

These gradients are used to update the model parameters:

$$
w_{new} = w_{old} - \alpha \frac{\partial L}{\partial w}
$$

$$
b_{new} = b_{old} - \alpha \frac{\partial L}{\partial b}
$$

where:

- $w$ = weight  
- $b$ = bias  
- $\alpha$ = learning rate  
- $L$ = loss  

---

# 1. Model Equation

Our neural network has only one dense neuron:

$$
\hat{y} = wx + b
$$

where:

- $x$ = input value  
- $w$ = weight  
- $b$ = bias  
- $\hat{y}$ = predicted value  

At **Epoch 1**, we start with:

$$
w = 0
$$

$$
b = 0
$$

So the prediction becomes:

$$
\hat{y} = 0 \times x + 0
$$

$$
\hat{y} = 0
$$


---

# 2. Loss Equation for One Sample

For one sample, the squared error loss is:

$$
L = (\hat{y} - y)^2
$$

Let:

$$
e = \hat{y} - y
$$

So:

$$
L = e^2
$$

where:

- $e$ = error  
- $\hat{y}$ = predicted value  
- $y$ = true value  

---

# 3. Derivative of Loss with Respect to Error

We have:

$$
L = e^2
$$

Differentiate with respect to $e$:

$$
\frac{\partial L}{\partial e} = 2e
$$

Since:

$$
e = \hat{y} - y
$$

we can write:

$$
\frac{\partial L}{\partial e} = 2(\hat{y} - y)
$$

---

# 4. Derivative of Error with Respect to Prediction

The error is:

$$
e = \hat{y} - y
$$

Differentiate with respect to $\hat{y}$:

$$
\frac{\partial e}{\partial \hat{y}} = 1
$$

Because:

$$
\frac{\partial}{\partial \hat{y}}(\hat{y}) = 1
$$

and:

$$
\frac{\partial}{\partial \hat{y}}(y) = 0
$$

So:

$$
\frac{\partial e}{\partial \hat{y}} = 1
$$

---
# 5. Derivative of Prediction with Respect to Weight and Bias

The prediction equation is:

$$
\hat{y} = wx + b
$$

Now we calculate the derivative of $\hat{y}$ with respect to both parameters:

$$
w
$$

and

$$
b
$$

# 6. Apply the Chain Rule for Weight and Bias

The dependency path is:

$$
parameter \rightarrow \hat{y} \rightarrow e \rightarrow L
$$

So the chain rule is:

<tr>
<td width="50%" valign="top">

## Weight Side

We want:

$$
\frac{\partial L}{\partial w}
$$

Apply chain rule:

$$
\frac{\partial L}{\partial w}
=
\frac{\partial L}{\partial e}
\times
\frac{\partial e}{\partial \hat{y}}
\times
\frac{\partial \hat{y}}{\partial w}
$$

$$
\frac{\partial L}{\partial e}=2e
$$

$$
\frac{\partial e}{\partial \hat{y}}=1
$$

$$
\frac{\partial \hat{y}}{\partial w}=x
$$

$$
\frac{\partial L}{\partial w}
=
2e \times 1 \times x
$$

$$
\frac{\partial L}{\partial w}
=
2ex
$$

Since:

$$
e=\hat{y}-y
$$

therefore:

$$
\boxed{
\frac{\partial L}{\partial w}

# 7. Weight and Bias Gradients for MSE with Many Samples

For many samples, we use Mean Squared Error:

$$
L =
\frac{1}{N}
\sum_{i=1}^{N}
(\hat{y}_i-y_i)^2
$$

and:

$$
error_i = \hat{y}_i-y_i
$$

<td width="50%" valign="top">

For one sample:

$$
\frac{\partial L}{\partial w}

2(\hat{y}-y)x

$$
\frac{\partial L}{\partial w}
=
\frac{1}{N}
\sum_{i=1}^{N}
2(\hat{y}_i-y_i)x_i
$$

$$
\frac{\partial L}{\partial w}

\frac{2}{N}
\sum_{i=1}^{N}
(\hat{y}_i-y_i)x_i
$$

$$

$$

the final formula is:

```python
import tensorflow as tf
import numpy as np
import pandas as pd

# -----------------------------
# Data
# -----------------------------
x_np = np.array([-40, -10, 0, 8, 15, 22, 38], dtype=np.float32).reshape(-1, 1)
y_np = np.array([-40.0, 14.0, 32.0, 46.4, 59.0, 71.6, 100.4], dtype=np.float32).reshape(-1, 1)



x = tf.constant(x_np)
y = tf.constant(y_np)

N = len(x_np)


w = tf.Variable([[0]], dtype=tf.float32)
b = tf.Variable([0], dtype=tf.float32)


learning_rate = 0.0001
epochs = 10


all_epoch_rows = []

# -----------------------------
# Training loop
# -----------------------------
for epoch in range(1, epochs + 1):

    with tf.GradientTape() as tape:
        y_pred = tf.matmul(x, w) + b
        error = y_pred - y
        loss = tf.reduce_mean(tf.square(error))

    dw, db = tape.gradient(loss, [w, b])

    # Save old values before update
    old_w = w.numpy()[0, 0]
    old_b = b.numpy()[0]
    loss_value = loss.numpy()
    dw_value = dw.numpy()[0, 0]
    db_value = db.numpy()[0]

    # Manual update
    new_w = old_w - learning_rate * dw_value
    new_b = old_b - learning_rate * db_value

    # Apply update
    w.assign([[new_w]])
    b.assign([new_b])

    # Save summary row
    all_epoch_rows.append({
        "epoch": epoch,
        "w_before": old_w,
        "b_before": old_b,
        "loss": loss_value,
        "dL_dw": dw_value,
        "dL_db": db_value,
        "w_after": new_w,
        "b_after": new_b
    })

    # -----------------------------
    # First 3 epochs: full manual explanation
    # -----------------------------
    if epoch <= 3:
        print("\n" + "=" * 90)
        print(f"EPOCH {epoch} FULL BACKPROPAGATION CALCULATION")
        print("=" * 90)

        print("\nModel equation:")
        print("y_pred = w*x + b")

        print("\nLoss equation:")
        print("MSE = (1/N) * Σ(y_pred - y_true)^2")

        print("\nGradient equations:")
        print("dL/dw = (2/N) * Σ((y_pred - y_true) * x)")
        print("dL/db = (2/N) * Σ(y_pred - y_true)")

        print("\nCurrent parameters before update:")
        print(f"w = {old_w:.6f}")
        print(f"b = {old_b:.6f}")

        # Build detailed per-sample table
        detailed_rows = []

        sum_error_sq = 0.0
        sum_error_x = 0.0
        sum_error = 0.0

        for i in range(N):
            xi = float(x_np[i, 0])
            yi = float(y_np[i, 0])

            pred_i = old_w * xi + old_b
            err_i = pred_i - yi
            err_sq_i = err_i ** 2
            err_x_i = err_i * xi

            sum_error_sq += err_sq_i
            sum_error_x += err_x_i
            sum_error += err_i

            detailed_rows.append({
                "i": i + 1,
                "x": xi,
                "y_true": yi,
                "y_pred = w*x+b": pred_i,
                "error = y_pred-y": err_i,
                "error^2": err_sq_i,
                "error*x": err_x_i
            })

        df_detail = pd.DataFrame(detailed_rows)

        print("\nPer-sample forward calculation:")
        print(df_detail.to_string(index=False))

        manual_loss = sum_error_sq / N
        manual_dw = (2 / N) * sum_error_x
        manual_db = (2 / N) * sum_error

        print("\nManual loss calculation:")
        print(f"Σ(error^2) = {sum_error_sq:.6f}")
        print(f"MSE = Σ(error^2) / N = {sum_error_sq:.6f} / {N}")
        print(f"MSE = {manual_loss:.6f}")

        print("\nManual gradient calculation for weight:")
        print(f"Σ(error*x) = {sum_error_x:.6f}")
        print(f"dL/dw = (2/N) * Σ(error*x)")
        print(f"dL/dw = (2/{N}) * {sum_error_x:.6f}")
        print(f"dL/dw = {manual_dw:.6f}")

        print("\nManual gradient calculation for bias:")
        print(f"Σ(error) = {sum_error:.6f}")
        print(f"dL/db = (2/N) * Σ(error)")
        print(f"dL/db = (2/{N}) * {sum_error:.6f}")
        print(f"dL/db = {manual_db:.6f}")

        print("\nParameter update equations:")
        print("w_new = w_old - learning_rate * dL/dw")
        print("b_new = b_old - learning_rate * dL/db")

        print("\nManual parameter update:")
        print(f"w_new = {old_w:.6f} - {learning_rate} * ({manual_dw:.6f})")
        print(f"w_new = {new_w:.6f}")

        print(f"b_new = {old_b:.6f} - {learning_rate} * ({manual_db:.6f})")
        print(f"b_new = {new_b:.6f}")

# -----------------------------
# Final table for all epochs
# -----------------------------
df_epochs = pd.DataFrame(all_epoch_rows)

print("\n" + "=" * 90)
print("SUMMARY TABLE FOR ALL EPOCHS")
print("=" * 90)

print(df_epochs.to_string(index=False))

print("\nFinal learned parameters:")
print(f"w = {w.numpy()[0, 0]:.6f}")
print(f"b = {b.numpy()[0]:.6f}")
```
